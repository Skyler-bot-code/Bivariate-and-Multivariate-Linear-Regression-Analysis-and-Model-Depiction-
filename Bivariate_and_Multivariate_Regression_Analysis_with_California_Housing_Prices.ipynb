## Bivariate Linear Regression ##

Linear Regression is considered the most common statistic technique to find relationships between variables and make predictions. With this being said, the goal of linear regression is to find a dependent variable that changes when one or more independent variable changes. Regression models captures casuality instead of degree of connections, meaning the regression model only shows cause and effect. Through out these relationships, researchers can better understand patterns in data to make predictions. There are numerous applications of linear regression models: businesses make prediction on future sales from advertising expenditures, engineers predict systems' performances based on different parameters, and data scientists confirm certain outcomes from observed patterns.

Linear regression model can be represent by the mathematical equation:
$$\large\hat{y} = \hat{b}_0 + \hat{m}_1x_1+\hat{m}_2x_2+\cdots+\hat{e}$$

where:
* $\hat{y}$ = The predicted value of the dependent variable

* $\hat{b}_0$ = The y intercept of the model, representing the expected value from the dependent variable when the independent variables equal zero

* $\hat{m}_1$ = The slope for $x_1$

* $\hat{m}_2$ = The slope for $x_2$

* $x_1$ = The first independent variable

* $x_2$ = The second independent variable

* $\hat{e}$ = The random error from the model


The above mathematical equation depicts multivariate linear regression, which also shows Ordinary Least Square (OLS) regression. OLS regression estimates the model's parameters by minimizing the sum of squared residuals. Residuals are the differences between marked values and the values predicted by the regression model. In conclusion, OLS regression model identifies the line that best fit the observed data.


The focus of the regression model explained in this article is bivariate linear regression, meaning the model examines relationships between one independent variable with one dependent variable. The mathematical form of bivariate linear regression is:

$$\large\hat{y} = \hat{b}_0 + \hat{m}x+\hat{e}$$


Linear regression provides measurements such as R-squared, p-values, confidence intervals, and residuals' diagnostics, allowing researchers to evaluate the reliability and quality of the model.





The purpose of the regression model is to predict the coefficients of values from the model that produce the best fitting line.

Ordinary Least Square from earlier explanation is a statistical principle of finding the line that minimizes the sum of squared residuals to predict the best fit line, which the mathematical equation for sum of squared errors (SSE) is:
$$\large\sum_{i=1}^{n} (y_i - \hat{y}_i)^2$$


Where:
* $\sum$ = Summation (add all values together)
* n = Total number of observations or data points

* $y_i$ = The actual (observed) value
* $\hat{y}_i$ = The predicted value (estimated by your regression line)

Several algorithms to minimize the Sum of Square errors include Normal Equation, Gradient Descent, and Stochastic Gradient Descent.

One algorithm to approach the coefficients for SSE is normal equation, which is a closed-form solution to find the optimal coefficients of a linear regression model without using an iterative optimization algorithm. The advantage of this algorithm is that the model gives exact solution without iteration to optimize the model. However, major disadvantages include highly unstable matrix inversion when parameters are highly correlated.

The mathematical equation is:

$$\large\hat{\boldsymbol{\beta}} = (X^T X)^{-1} X^T \mathbf{y}$$

where:

*   $\hat{\boldsymbol{\beta}}$ is the vector of coefficients (including the intercept).
*   $X$ is the design matrix (features, with a column of ones for the intercept).
*   $X^T$ is the transpose of the design matrix.
*   $(X^T X)^{-1}$ is the inverse of the product of $X^T$ and $X$.
*   $\mathbf{y}$ is the vector of observed dependent variable values.

Next algorthim is the gradient descent algorithm. Gradient descent is an iterative optimization alogorithm that finds the minimun of a function. In the context of linear regression, the aim is to minimize the cost function, which is the Mean Squared Error. Through continuous iterations, it adjusts the model's parameters (coefficients) that is in the opposite direction to the gradient of the cost function. The advantages of using the model is that it is efficient for large datasets and suitable for high dimentional problems by avoiding matix inversions. However, the disadvantages are that it is time consuming and can be not accurate when parameters are chosen poorly.

The rule for parameters in the gradient descent algorithm can be compute through the below mathematical equation:

$$\large\theta_{j} := \theta_{j} - \alpha \frac{\partial}{\partial \theta_{j}} J(\theta)$$

where:
*   $\theta_{j}$ is the $j$-th parameter (coefficient) of the model.
*   $J(\theta)$ is the cost function (e.g., Mean Squared Error).
*   $\alpha$ is the learning rate, which determines the size of the steps taken during each iteration.
*   $\frac{\partial}{\partial \theta_{j}} J(\theta)$ is the partial derivative of the cost function with respect to the parameter $\theta_{j}$, which indicates the direction of the steepest ascent.

Another specific gradient algorithm is called the Stochastic gradient descent algorithm (SGD). Instead of computing the cost function for the entire model in normal gradient descent model, Stochastic gradient descent algorithm computes cost function and updates parameters for each training examples individually, making it a faster algorithms for large datasets. However, the disadvantages are that it requires more careful tunning and more variable convergence path.

The mathematical rule for parameters in Stochastic gradient descent is:

$$\large\theta_{j} := \theta_{j} - \alpha (h_\theta(x^{(i)}) - y^{(i)}) x_{j}^{(i)}$$

where:
*   $\theta_{j}$ is the $j$-th parameter (coefficient) of the model.
*   $\alpha$ is the learning rate, which determines the step size for each update.
*   $h_\theta(x^{(i)})$ is the predicted output of the model for the $i$-th training example $x^{(i)}$, given the current parameters $\theta$.
*   $y^{(i)}$ is the actual (true) output for the $i$-th training example $x^{(i)}$.
*   $x_{j}^{(i)}$ is the $j$-th feature of the $i$-th training example. This is used to scale the error in the direction of the $j$-th parameter.

In conclusion, in linear regression models, Gradient descent and Stochastic gradient descent methods are the most suitable when analyzing from large datasets. However, since the dataset is medium, using OLS model with `statsmodels.ols(formula, data = dataframe).fit()`, which is the most simlar to a Normal Equation Method, would be the best approach.

Before each iteration, we will define $H_0$ and $H_a$ for hypothesis testing. Essentially, we are determining whether the four independent variables have explanatory power of the dependent variable.

## For each iteration: ##

Null Hypothesis ($H_0$): For each iteration, the independent variables have no linear relationship with the median_house_value, meaning the regression coefficient for the slope is zero.

 $$\large H_0: \beta_i = 0$$

Alternative Hypothesis ($H_a$): The independent variables have a relationship with the median_house_value, meaning the coefficient for the slope is not zero

 $$\large H_a: \beta_i \ne 0$$


The first relationship being discussed is median house value with longtitude, showing a negative relationship. To visualize the prediction of a linear equation for the relationship, we will graph a linear regression line with two lines above and below 2 $\times$ standard deviation, and include data points to see the significant outliers. We set the outliers to be the mean + 2 $\times$  standard deviation because 95% of the data points in normal distribution fall within two standard deviations.

The mathematical equation of the bivariate linear regression is given:

$$\large \text{Median House Value} = \hat{b}_0 + \hat{m}\text{Longitude}+\hat{e}$$



```
import plotly.graph_objects as go
import numpy as np
import statsmodels.formula.api as smf

# Bivariate linear regression with outliers

formula = "Q('median_house_value') ~ Q('longitude')" # Corrected 'longtitude' to 'longitude'
model = smf.ols(formula, data = df_house)
reg_results_1 = model.fit()

# Values of coefficent b and m of the model
[intercept, slope] = reg_results_1.params
print(f"intercept = {intercept: .3f} and the slope = {slope: .3f}")

# Fitted value
yhat = reg_results_1.fittedvalues

# residuals (Errors)
ehat = reg_results_1.resid

# Sum of squared errors
sse = sum(np.square(ehat))

# Standard Deviations
std = np.sqrt(sse / (len(ehat) - 2))



# Plot the linear regression line and mean of y + 2std above and below



fig = go.Figure() # opening a plotting area

# add a trace for the regression line
fig.add_trace(go.Scatter(
                        x = df_house["longitude"], # x = predictor
                        y = df_house["median_house_value"], # y = resond
                        mode = "markers",
                        name = "Actual",
                        marker = dict(color = "blue",
                                      symbol = "square-open",
                                      size = abs(ehat)/100000,
                                      sizemin = 4
                                      ),
                        hovertext = df_house.index,
                        error_y = dict(type = "data",
                                        symmetric = False,
                                        array = -ehat,
                                        color = "lightgreen",
                                        thickness = 1
                                        )

                        ))

fig.add_trace(go.Scatter(
                        x = df_house["longitude"], # x = predictor
                        y = yhat, # y = respond
                        mode = "lines",
                        name = "Predicted"

                        ))

# draw the boundaries for significant outliers

# mean of y + 2 * std above
fig.add_trace(go.Scatter(
                        x = df_house["longitude"], # x = predictor
                        y = yhat + 2 * std, # y = respond
                        mode = "lines",
                        line = dict(color = "orange", width = 2, dash = "dash"),
                        name = "Predicted + 2std"

                        ))
# mean of y + 2 * std below
fig.add_trace(go.Scatter(
                        x = df_house["longitude"], # x = predictor
                        y = yhat - 2 * std, # y = respond
                        mode = "lines",
                        line = dict(color = "orange", width = 2, dash = "dash"),
                        name = "Predicted - 2std"

                        ))


fig.update_layout(
                  height = 800, width = 1000,
                  title = f"Equation: median house value = {slope: .3f}(longitude) + {intercept: .3f}", # Corrected title as well
                  title_x = 0.5,
                  xaxis_title = "longitude", # Corrected title as well
                  yaxis_title = "Median House Value",
                  font= dict(color = "green", size = 20)
                  )
```

Below is a Oridnary Least Squares Regression Results, it provides a comprehensive overview of the regression model's performance.

```
# OLS Regression Results for Median House Value and Longitude
print(reg_results_1.summary())
```

The p-value of the intercept and the slope are greater than $\alpha = 0.05$, indicating that both coefficients are insignificant from the model. Moreover, the adjusted $R^2$ is negative, suggesting that the linear regression model cannot explain the relationship between median household value and longtitude. Furthermore, the F-statistic value is 0.05253, and its p-value is 0.819 > $\alpha = 0.05$, indicating that the data fails to reject the null hypothesis that there is no relation between the median house value and median income, or the slope = 0 as the regression is a horizontal line. In other words, the model (slope) is instatistically significant.

By evaluating the model assumptions using durbin-watson statistic, the value is 2.114. A value around 2 suggests no significant first-order serial correlation in the residuals. This indicates that the assumption of independent errors is likely met.

By evaluating the normality of residuals using Jarque-Bera Test, the p-value is $9.64 \cdot 10^{-10} <<<<< \alpha = 0.05$. We reject the null hypothesis that the residuals are normally distributed. This suggests that the residuals of this model are not normally distributed.

## Multivariate Linear Regression ##

$$\large\text{median_house_value} = \hat{b}_0+\hat{m}_1\text{median_income} + \hat{m}_2\text{housing_median_age}+e$$

```
# Perform Multivariate Linear Regression
import statsmodels.formula.api as smf
import numpy as np

formula_multi = "Q('median_house_value') ~ Q('median_income') + Q('housing_median_age')"
model_multi = smf.ols(formula_multi, data = df_house)
reg_results_multi = model_multi.fit()

print(reg_results_multi.summary())

# Get coefficients and other metrics from the multivariate model
coef_multi = reg_results_multi.params
ehat_multi = reg_results_multi.resid
sse_multi = sum(np.square(ehat_multi))
std_multi = np.sqrt(sse_multi / (len(ehat_multi) - len(coef_multi)))

print(f"Multivariate model intercept = {coef_multi[0]: .3f}")
print(f"Coefficient for median income (%) = {coef_multi[1]: .3f}")
print(f"Coefficient for housing median age (%) = {coef_multi[2]: .3f}")
print(f"Standard deviation of residuals = {std_multi: .3f}")
```

```
# Multivariate Regression Analysis
import plotly.graph_objects as go
import numpy as np

# Coefficients from the multivariate model
intercept_m = coef_multi[0]
slope_1 = coef_multi[1]
slope_2 = coef_multi[2]

print(f"The intercept is {intercept_m: .1f}")

fig = go.Figure()

# create a 2D grid for the independent variables
x_range = np.linspace(df_house["median_income"].min(), df_house["median_income"].max(), 100)
y_range = np.linspace(df_house["housing_median_age"].min(), df_house["housing_median_age"].max(), 100)

X_grid, Y_grid = np.meshgrid(x_range, y_range)

# create the coordinate for a 3d plane using multivariate coefficients
Z_plane = slope_1 * X_grid + slope_2 * Y_grid + intercept_m

# Add actual data points
fig.add_trace(
              go.Scatter3d(
                            x = df_house["median_income"],
                            y = df_house["housing_median_age"],
                            z = df_house["median_house_value"],
                            mode = "markers",
                            marker = dict(size = 4, color = "red", opacity = 0.5),
                            name = "Actual Data"
                          )
              )

# Add regression plane
fig.add_trace(go.Surface(
                         x = X_grid, y = Y_grid, z = Z_plane, colorscale = "blues", showscale = False,
                         name = "Regression Plane",
                         opacity = 0.6
                         )
              )

# Add plane for + 2 * std_multi
fig.add_trace(go.Surface(
                         x = X_grid, y = Y_grid, z = Z_plane + 2 * std_multi, colorscale = "viridis", showscale = False,
                         name = "Predicted + 2 * Std",
                         opacity = 0.3
                         )
              )

# Add plane for - 2 * std_multi
fig.add_trace(go.Surface(
                         x = X_grid, y = Y_grid, z = Z_plane - 2 * std_multi, colorscale = "viridis", showscale = False,
                         name = "Predicted - 2 * Std",
                         opacity = 0.3
                         )
              )

fig.update_layout(width = 1000, height = 700,
                  title = f"Equation: median house value = {slope_1: .3f}(median income) + {slope_2: .3f}(housing median age) + {intercept_m: .3f}",
                  title_x = 0.5,
                  scene = dict(
                               xaxis_title = "median income",
                               yaxis_title = "housing median age",
                               zaxis_title = "median house value"
                               )
                  )

fig.show()
```

```
import statsmodels.stats.api as sms
from statsmodels.compat import lzip

name = ["Lagrange multiplier statistic", "p-value", "f-value", "p-value for f-value"]
test = sms.het_breuschpagan(ehat1, reg_results_no_outliers.model.exog)

lzip(name, test)
```

Based on the above result, the model fails to rejects the null hypothesis to some extent because not all p-value are greater than $\alpha = 0.05$. Therefore, this indicates that heteroscedasticity is present in the residuals model. In other words, it suggests that the variance of an error is not constant over the entire levels of median income, violating one key assumptions when using ordinary least squared (OLS) regression.

Conclusion of the reliability of the regression model:

The final bivariate linear regression model, which predicts median house value using median income(without outliers and without an intercept), has a strong explanatory power. It achieves a very high uncentered Adjusted R-squared of 0.914 (a significant proportion of the variance), and the Jarque-Bera test showed that the residuals are normally distributed (p-value > 0.05).

However, the visual diagnostic plots and the formal way of Breusch-Pagn statistical test both strongly show the presence of heteroscedasticity, which is that the variance of the errors is not constant across all the levels of median income.

Therefore, the model is not reliable enough for vigorous statistical inferences.
